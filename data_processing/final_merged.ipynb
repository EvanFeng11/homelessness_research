

**Merging Variables**

- `output_data/LAD_landuse_group_stats_wide.csv`  
- `output_data/LAD_free_POI_group_with_9_category_summary_wide.csv`  
- `output_data/merged_census_data.csv`  
- `output_data/cleaned_crime_total_sum.csv`  
- `output_data/cleaned_crime_asb_sum.csv`  

The dependent variable organized in Excel: `Detailed_LA_20222023_A1.csv`

Finally merged into `output_data/final_merged.csv`.

This version contains some irrelevant variables and many empty fields. A cleaned version, with field names organized and irrelevant variables and empty fields removed, is saved as `output_data/final_merged_cleaned.csv`.

This CSV file was later saved in `"Homelessness_England/analysis/final_merged_cleaned.csv"`, and the two files are identical.

```python
import pandas as pd

# Read files
df_landuse = pd.read_csv('output_data/LAD_landuse_group_stats_wide.csv')
df_socio = pd.read_csv('output_data/merged_census_data.csv')
df_poi = pd.read_csv('output_data/LAD_free_POI_group_with_9_category_summary_wide.csv')

# Standardize the primary key field name
df_socio = df_socio.rename(columns={'geography code': 'LAD21CD'})

# Merge dataframes
df_merged = df_landuse.merge(df_socio, on='LAD21CD').merge(df_poi, on='LAD21CD')

# View results
print(df_merged.head())

#df_merged.to_csv('merged_output_01.csv', index=False)
```

```python
# Merge dependent variables

# Read the already merged main table
df_main = df_merged

# Read the detailed table
df_detail = pd.read_csv('Detailed_LA_20222023_A1.csv')

# Check the primary key field name of the detailed table (assumed to be "LAD21CD", replace if different)
# For example, if it is 'code', use df_detail.rename(columns={'code': 'LAD21CD'}, inplace=True)

# Merge
df_homeless = df_main.merge(df_detail, on='LAD21CD', how='left')
```

```python
# Merge crime data

# Read the main table
df_main2 = df_homeless

# Read crime data
df_asb = pd.read_csv('output_data/cleaned_crime_asb_sum.csv')
df_total = pd.read_csv('output_data/cleaned_crime_total_sum.csv')

# Rename primary key field
df_asb = df_asb.rename(columns={'Local Authority code': 'LAD21CD'})
df_total = df_total.rename(columns={'Local Authority code': 'LAD21CD'})

# Merge ASB crime data
df_merge1 = df_main2.merge(df_asb, on='LAD21CD', how='left')

# Merge total crime data
df_final = df_merge1.merge(df_total, on='LAD21CD', how='left')

# Save the results
df_final.to_csv('output_data/final_merged.csv', index=False)
```
